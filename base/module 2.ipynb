### list

```python
nums = [10, 20, 30]

print(nums[0])   # 10，下标从 0 开始
print(nums[-1])  # 30，-1 表示倒数第一个

# 切片：列表[起点:终点:步长]，包含起点，不包含终点
print(nums[1:3])  # [20, 30]，取下标 1、2
print(nums[:2])   # [10, 20]，省略起点，从头开始
print(nums[1:])   # [20, 30]，省略终点，取到末尾
print(nums[:])    # [10, 20, 30]，复制整个列表

# nums[::-1] 表示从后往前取，得到反转后的新列表
reversed_nums = nums[::-1]
print(reversed_nums)  # [30, 20, 10]
print(nums)           # [10, 20, 30]，原列表没有改变
```

```python
nums.append(40)         # 末尾添加，均摊 O(1)
nums.extend([50, 60])   # 批量添加
last = nums.pop()       # 删除末尾，O(1)
first = nums.pop(0)     # 删除开头，O(n)
nums.insert(1, 99)      # 中间插入，O(n)
nums.remove(30)         # 查找并删除第一个 30，O(n)
30 in nums              # 存在性检查，O(n)
```

### dict 字典

```python
# 计数
nums = [2, 3, 2, 4, 3, 2]

frequency = {}
for value in nums:
    frequency[value] = frequency.get(value, 0) + 1
    # frequency.get(value, 0)：查询 value 已经出现了几次；如果字典里没有这个键，就返回 0

print(frequency)  # {2: 3, 3: 2, 4: 1}

# 分组
words = ["eat", "tea", "tan", "ate", "nat"]

groups = {}
for word in words:
    key = "".join(sorted(word))
    groups.setdefault(key, []).append(word)

print(groups)
# {'aet': ['eat', 'tea', 'ate'], 'ant': ['tan', 'nat']}

# 记录最后出现位置
nums = [2, 3, 2, 4, 3]

last_position = {}
for index, value in enumerate(nums):
    last_position[value] = index

print(last_position)  # {2: 2, 3: 4, 4: 3}
```

### set 集合

```python
seen = set()
seen.add(3)
3 in seen               # 平均 O(1)
seen.discard(9)         # 不存在也不会报错
```

```python
# 用 dict 统计数组频率，并输出频率最高的元素。
nums = [2, 3, 2, 4, 3, 2]
frequency = {}

for value in nums:
    frequency[value] = frequency.get(value,0)+1

print(frequency)
```

```python
# 用 set 判断数组是否存在重复元素
def containsDuplicate(nums):
    seen = set()

    for value in nums:
        if value in seen:
            return True
        seen.add(value)

    return False

print(containsDuplicate([1, 2, 3, 1]))  # True
print(containsDuplicate([1, 2, 3, 4]))  # False
print(containsDuplicate([]))           # False
```

```python
# 用 defaultdict(list) 按字符串长度分组
# defaultdict(list) 会在访问不存在的键时，自动创建一个空列表，方便直接用 append() 添加元素。
from collections import defaultdict

words = ["a", "to", "cat", "hi", "dog", "four"]

groups = defaultdict(list)

for word in words:
    groups[len(word)].append(word)

print(dict(groups))
```

```python
class ListNode:
    def __init__(self, value):
        self.value = value
        self.next = None


# 构造链表 1 -> 2 -> 3
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)

# 遍历链表
current = head
while current is not None:
    print(current.value)
    current = current.next
```

```python
# 构造一棵三层二叉树，写函数统计节点数
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None


# 构造三层二叉树：
#       1
#      / \
#     2   3
#    / \   \
#   4   5   6

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.right = TreeNode(6)


# 递归统计节点数
def count_nodes(node):
    if node is None:
        return 0
    return 1 + count_nodes(node.left) + count_nodes(node.right)

print(count_nodes(root))  # 6
print(count_nodes(None))  # 0
```

```python
# 实现一个支持 add 和 size 的简单集合类
class MySet:
    def __init__(self):
        self.data = set()  # 每个实例都有自己的集合

    def add(self, value):
        self.data.add(value)

    def size(self):
        return len(self.data)


s = MySet()
s.add(1)
s.add(2)
s.add(1)  # 重复元素不会再次添加

print(s.size())  # 2
```
