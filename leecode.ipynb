### 1.两数之和
哈希表-记忆功能

```python
def twoSum(nums, target):
    hashmap = {}
    for idx, num in enumerate(nums):
        complement = target - num
        if complement in hashmap:
            return [hashmap[complement], idx]
        hashmap[num] = idx
    return []

print(twoSum([2,7,11,15],9)) # [0,1]
print(twoSum([3,2,4],6))     # [1,2]
print(twoSum([3,3],6))       # [0,1]
```

### 167.两数之和2
输入有序数组（非递减）- 双指针

```python
def twoSum(numbers, target):
    left = 0
    right = len(numbers) - 1
    while left < right:
        s = numbers[left] + numbers[right]
        if s == target:
            return [left+1, right+1]
        elif s < target:
            left += 1
        else:
            right -= 1
    return []

print(twoSum([2,7,11,15],9)) # [1,2]
print(twoSum([2,3,4],6))     # [1,3]
print(twoSum([-1,0],-1))     # [1,2]
```

### 209. 长度最小的子数组
滑动窗口：全正/全负，子数组

```python
def minSubArrayLen(target, nums):
    left = 0
    s = 0
    min_len = len(nums) + 1

    for right in range(len(nums)):
        s += nums[right]

        while s >= target:
            min_len = min(min_len, right - left + 1)
            s -= nums[left]
            left += 1

    return min_len if min_len <= len(nums) else 0


print(minSubArrayLen(7, [2, 3, 1, 2, 4, 3]))         # 2
print(minSubArrayLen(4, [1, 4, 4]))                  # 1
print(minSubArrayLen(11, [1, 1, 1, 1, 1, 1, 1, 1]))  # 0
```

```python
def minSubArrayLen(target, nums):
    n = len(nums)

    prefix = [0]
    for value in nums:
        prefix.append(prefix[-1] + value)

    min_len = n + 1

    for left in range(n):
        goal = prefix[left] + target

        # 左闭右开二分查找
        low = left + 1
        high = n + 1

        while low < high:
            mid = (low + high) // 2

            if prefix[mid] >= goal:
                high = mid
            else:
                low = mid + 1

        if low <= n:
            min_len = min(min_len, low - left)

    return min_len if min_len <= n else 0


print(minSubArrayLen(7, [2, 3, 1, 2, 4, 3]))  # 2
print(minSubArrayLen(4, [1, 4, 4]))           # 1
print(minSubArrayLen(11, [1] * 8))           # 0
```

### 3. 无重复字符的最长子串
集合、滑动窗口

输入: s = "pwwkew"

输出: 3

解释: 请注意，"pwke" 是一个子序列，不是子串。子串必须连续，子序列可以跳过字符，但不能改变先后顺序

```python
def lengthOfLongestSubstring(s):
    left = 0
    seen = set()
    max_len = 0

    for right in range(len(s)):
        while s[right] in seen:    # remove直到无重复元素，不可以用if，否则只去掉一个元素
            seen.remove(s[left])
            left += 1

        seen.add(s[right])
        max_len = max(max_len, right - left + 1)

    return max_len


print(lengthOfLongestSubstring("abcabcbb"))  # 3
print(lengthOfLongestSubstring("bbbbb"))     # 1
print(lengthOfLongestSubstring("pwwkew"))    # 3
```

```python
def lengthOfLongestSubstring(s):
    left = 0
    right = 0
    seen = set()
    max_len = 0

    while right < len(s):
        if s[right] in seen:
            seen.remove(s[left])
            left += 1
        else:
            seen.add(s[right])
            max_len = max(max_len, len(seen))
            right += 1    # 关键是 right += 1 只放在 else 里面

    return max_len

print(lengthOfLongestSubstring("abcabcbb"))  # 3
print(lengthOfLongestSubstring("bbbbb"))     # 1
print(lengthOfLongestSubstring("pwwkew"))    # 3
```

### 26. 删除有序数组中的重复项
读/写指针、快速排序、双指针复杂度

```python
def removeDuplicates(nums):
    slow = 0  # 第一个元素直接保留

    for fast in range(1, len(nums)):
        if nums[fast] != nums[slow]:
            nums[slow+1] = nums[fast]
            slow += 1

    return slow + 1


nums = [1, 1, 2]
k = removeDuplicates(nums)
print(k)         # 2
print(nums[:k])  # [1, 2]

nums = [0, 0, 1, 1, 1, 2, 2, 3, 3, 4]
k = removeDuplicates(nums)
print(k)         # 5
print(nums[:k])  # [0, 1, 2, 3, 4]
```

### 704. 二分查找
logN复杂度

```python
def search(nums, target):
    left = 0
    right = len(nums)
    while left < right:
        mid = (left + right) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid
    return -1

print(search([-1, 0, 3, 5, 9, 12], 9))  # 4
print(search([-1, 0, 3, 5, 9, 12], 2))  # -1
    
```

### 59. 螺旋矩阵 II
边界模拟、代码调试、撞南墙

```python
# 常规法
def generateMatrix(n):
    matrix = [[0] * n for _ in range(n)]
    count = 1
    start = 0

    while count < n**2:
        # 向右
        for j in range(start, n - 1 - start):
            matrix[start][j] = count
            count += 1

        # 向下
        for i in range(start, n - 1 - start):
            matrix[i][n - 1 - start] = count
            count += 1

        # 向左
        for j in range(n - 1 - start, start, -1):
            matrix[n - 1 - start][j] = count
            count += 1

        # 向上
        for i in range(n - 1 - start, start, -1):
            matrix[i][start] = count
            count += 1

        start += 1

    # 奇数阶矩阵还剩一个中心格
    if n % 2 == 1:
        matrix[start][start] = count

    return matrix


print(generateMatrix(1))  # [[1]]
print(generateMatrix(2))  # [[1, 2], [4, 3]]
print(generateMatrix(3))  # [[1, 2, 3], [8, 9, 4], [7, 6, 5]]
```

```python
# 边界收缩法（撞南墙）
def generateMatrix(n):
    matrix = [[0] * n for _ in range(n)]
    top, bottom, left, right = 0, n-1, 0, n-1
    count = 1
    while count <= n**2:
        for j in range(left, right+1):
            matrix[top][j] = count
            count += 1
        top += 1

        for i in range(top, bottom+1):
            matrix[i][right] = count
            count += 1
        right -= 1

        for j in range(right, left-1, -1):
            matrix[bottom][j] = count
            count += 1
        bottom -= 1

        for i in range(bottom, top-1, -1):
            matrix[i][left] = count
            count += 1
        left += 1

    return matrix

print(generateMatrix(1))  # [[1]]
print(generateMatrix(2))  # [[1, 2], [4, 3]]
print(generateMatrix(3))  # [[1, 2, 3], [8, 9, 4], [7, 6, 5]]
```

### 303. 区域和检索
面向对象简介、前缀和：子数组的和/平均

```python
class NumArray:
    def __init__(self, nums):
        self.prefix = [0]

        for value in nums:
            self.prefix.append(self.prefix[-1] + value)

    def sumRange(self, left, right):
        return self.prefix[right + 1] - self.prefix[left]


numArray = NumArray([-2, 0, 3, -5, 2, -1])

print(numArray.sumRange(0, 2))  # 1
print(numArray.sumRange(2, 5))  # -1
print(numArray.sumRange(0, 5))  # -3
```

```python
# 暴力解，时间复杂度O(n)
class NumArray1:

    def __init__(self, nums: List[int]):
        self.nums=nums

    def sumRange(self, left: int, right: int) -> int:
        my_sum=0
        for i in range(left,right+1):
            my_sum+=self.nums[i]

        return my_sum

# 前缀和另一种写法，时间复杂度：O(1)
class NumArray:

    def __init__(self, nums: List[int]):
        self.preSum = [0] * (len(nums) + 1)
        for i in range(len(nums)):
            self.preSum[i+1]=self.preSum[i]+nums[i]

    def sumRange(self, left: int, right: int) -> int:
        return self.preSum[right+1]-self.preSum[left]
```

### 560. 和为 K 的子数组
前缀和、两数之和（哈希表）综合应用

```python
class Solution(object):
    def subarraySum(self, nums, k):
        prefix = [0]
        for value in nums:
            prefix.append(prefix[-1] + value)

        hashmap = {}
        count = 0

        for value in prefix:
            complement = value - k

            if complement in hashmap:
                count += hashmap[complement]

            hashmap[value] = hashmap.get(value, 0) + 1

        return count


solution = Solution()
print(solution.subarraySum([1, 1, 1], 2))  # 2
print(solution.subarraySum([1, 2, 3], 3))  # 2
print(solution.subarraySum([0, 0], 0))     # 3
```

### 1109. 航班预订统计
差分数组：每条记录只修改两个位置，最后遍历一次还原

```python
def corpFlightBookings(bookings, n):
    answer = [0] * n

    for first, last, seats in bookings:
        answer[first - 1] += seats

        if last < n:
            answer[last] -= seats

    # 累加差分数组，还原每个航班的座位数
    for i in range(1, n):
        answer[i] += answer[i - 1]

    return answer


print(corpFlightBookings(
    [[1, 2, 10], [2, 3, 20], [2, 5, 25]], 5
))
# [10, 55, 45, 25, 25]

print(corpFlightBookings(
    [[1, 2, 10], [2, 2, 15]], 2
))
# [10, 25]
```
