### 206. 反转链表
原地反转

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


# 将列表转换为链表
def build_list(nums):
    dummy = ListNode()
    cur = dummy

    for num in nums:
        cur.next = ListNode(num)
        cur = cur.next

    return dummy.next


# 将链表转换为列表，方便查看结果
def to_list(head):
    result = []

    while head is not None:
        result.append(head.val)
        head = head.next

    return result


# 方法一：新建节点，头插法，保留原链表
def reverse_copy(head):
    cur = None

    while head != None:
        cur = ListNode(head.val, cur)
        head = head.next

    return cur


# 方法二：原地反转，修改原节点的 next
def reverse_in_place(head):
    pre = None

    while head != None:
        next_node = head.next
        head.next = pre
        pre = head
        head = next_node

    return pre


# 测试两种方法
for nums in [[1, 2, 3, 4, 5], [1, 2], [], [1]]:
    # 方法一
    head = build_list(nums)
    new_head = reverse_copy(head)

    print("输入：", nums)
    print("新建节点反转：", to_list(new_head))
    print("原链表：", to_list(head))  # 原链表不变

    # 方法二：重新构造一条链表进行测试
    head = build_list(nums)
    new_head = reverse_in_place(head)

    print("原地反转：", to_list(new_head))
    print()
```

### 21. 合并两个有序链表
dummy虚拟头节点

```python
class Solution(object):
    def mergeTwoLists(self, l1, l2):
        dummy = ListNode()
        cur = dummy

        while l1 != None and l2 != None:
            if l1.val <= l2.val:
                cur.next = l1
                l1 = l1.next
            else:
                cur.next = l2
                l2 = l2.next

            cur = cur.next

        # 将还没遍历完的链表接到末尾
        if l1 != None:
            cur.next = l1
        else:
            cur.next = l2

        return dummy.next
```

### 141. 环形链表
判断链表是否有环 - 快慢指针套圈法

```python
class Solution(object):
    def hasCycle(self, head):
        slow = head
        fast = head

        while fast != None and fast.next != None:
            slow = slow.next
            fast = fast.next.next

            if slow == fast:
                return True

        return False
```

```python
# 哈希表法
class Solution(object):
    def hasCycle(self, head):
        seen = set()

        while head is not None:
            if head in seen:
                return True

            seen.add(head)
            head = head.next

        return False
```

### 142. 环形链表Ⅱ

```python
class Solution(object):
    def detectCycle(self, head):
        slow = head
        fast = head

        # 第一阶段：判断是否有环
        while fast is not None and fast.next is not None:
            slow = slow.next
            fast = fast.next.next

            if slow is fast:
                # 第二阶段：寻找入口
                cur = head

                while cur is not slow:
                    cur = cur.next
                    slow = slow.next

                return cur

        return None
```
