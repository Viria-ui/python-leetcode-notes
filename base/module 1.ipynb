```python
nums = [10, 20, 30]

for value in nums:
    print(value)

for index, value in enumerate(nums):
    print(index, value)
    
# 反向遍历
for value in reversed(nums):
    print(value)

for index in range(len(nums) - 1, -1, -1):
    print(index, nums[index])
```

range(start, stop, step) 包含 start，不包含 stop。

遍历下标时使用 range(len(nums))，同时需要下标和值时优先用 enumerate(nums)。

```python
# 双指针
nums = [2,7,11,15]
target = 9
left = 0
right = len(nums) - 1

while left < right:
    if nums[left] + nums[right] < target:
        left += 1
    else:
        right -= 1
```

写 while 前回答三个问题：
循环开始前，哪些变量表示当前状态？
每一轮怎样让状态接近结束？
什么时候必须停止？

循环过程中始终成立的事实叫循环不变量。例如二分查找中，“答案如果存在，一定还在 [left, right] 范围内”就是不变量。

```python
# 二分查找
# 左闭右闭区间 [left, right]
middle = (left + right) // 2
if nums[middle] < target:
    left = middle + 1     # 不可以是 left = middle
else:
    right = middle - 1

# 左闭右开区间 [left, right)
middle = (left + right) // 2
if nums[middle] < target:
    left = middle + 1
else:
    right = middle
```

```python
# 不使用 max()，求一个非空列表的最大值
nums = [2, 7, 11, 15]
max_value = nums[0]

for value in nums:    # 直接用 nums 遍历元素
    if value > max_value:
        max_value = value

print(max_value)  # 15
```

```python
nums = [2,7,11,15]
target = 9
left = 0
right = len(nums) - 1
while left < right:
    s = nums[left] + nums[right]
    if s == target:
        print(left, right)
        break    # 跳出死循环
    elif s < target:
        left += 1
    else:
        right -= 1
```

```python
# count_even(nums) 返回偶数个数
def count_even(nums):
    count = 0
    for value in nums:
        if value % 2 == 0:
            count += 1
        return count
```

```python
# 写 divide(a, b)，当 b == 0 时返回 None
def divide(a, b):
    if b == 0:
        return None
    else:
        return a / b
```

```python
# 求列表元素和
# 用循环
def sum_list(nums):
    s = 0
    for value in nums:
        s += value
    return s

# 用递归
def list_sum(nums, index=0):
    if index == len(nums):
        return 0
    return nums[index] + list_sum(nums, index + 1)

print(list_sum([1, 2, 3, 4]))  # 10
print(list_sum([2, 7, 11, 15]))  # 35
print(list_sum([]))  # 0
```

这里在 sum_list 函数里面，又调用了 sum_list 自己，所以叫递归。
递归需要两点：
- 结束条件：到达列表末尾时，返回 0，不再调用自己。
- 逐步接近结束条件：每次把 index 加 1，处理下一个元素。

### 分别用返回值和 nonlocal 实现二叉树节点计数，比较可读性

1. 什么是二叉树？
二叉树由一个个“节点”组成，每个节点最多有两个子节点，分别叫左子节点和右子节点  
       1
      / \
     2   3
    /
   4

```python
class TreeNode:
    def __init__(self, value):
        self.value = value   # 节点保存的值
        self.left = None     # 左子节点，None 表示没有
        self.right = None    # 右子节点

root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
```

2. 用返回值计数
思路是：整棵树的节点数 = 1 个当前节点 + 左边的节点数 + 右边的节点数。

```python
def count_nodes(node):
    if node is None:
        return 0

    return 1 + count_nodes(node.left) + count_nodes(node.right)


print(count_nodes(root))  # 3，函数调用自己，分别计算左右两边，这就是递归。根节点最终返回 1 + 1 + 1 = 3
print(count_nodes(None))  # 0，空树没有节点
```

3. 什么是 nonlocal？
当一个函数里面又定义了一个函数，内层函数如果需要给外层函数的变量重新赋值，可以使用 nonlocal。

```python
def example():
    count = 0

    def add_one():
        nonlocal count # 表示这里使用的是外层函数的 count
        count += 1

    add_one()
    print(count)  # 1
```

```python
# 统计树的节点
def count_nodes_nonlocal(root):
    count = 0

    def visit(node):
        nonlocal count

        if node is None:
            return

        count += 1
        visit(node.left)
        visit(node.right)

    visit(root)
    return count
```

```python
class TreeNode:
    def __init__(self, value, left=None, right=None):
        self.value = value
        self.left = left
        self.right = right


# 方法一：用返回值
def count_nodes(root):
    if root is None:
        return 0

    return 1 + count_nodes(root.left) + count_nodes(root.right)

# 方法二：用 nonlocal
def count_nodes_nonlocal(root):
    count = 0

    def visit(node):
        nonlocal count

        if node is None:
            return

        count += 1
        visit(node.left)
        visit(node.right)

    visit(root)
    return count

# 构建二叉树：
#     1
#    / \
#   2   3
#  /
# 4
root = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))

print(count_nodes(root))           # 4
print(count_nodes_nonlocal(root))  # 4

# 空树测试
print(count_nodes(None))           # 0
print(count_nodes_nonlocal(None))  # 0
```

返回值版本把问题拆成：当前节点 1 + 左子树节点数 + 右子树节点数。每次递归都返回一个计数，代码短，含义直接，这道题更推荐这种写法。
nonlocal 版本每访问一个节点，就把外层的 count 加 1。nonlocal count 允许内层函数修改外层函数的变量。它也容易理解，但多了嵌套函数和共享计数器，需要关注变量在哪里被修改。
两种方法的时间复杂度都是 O(n)，递归栈空间都是 O(h)，其中 n 是节点数，h 是树的高度。
