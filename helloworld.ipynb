【问题】奇偶分家。给定一组非负整数，请统计奇数和偶数各有多少个？
【问题分析】
1）输入：一行空格分隔的非负整数
2）处理：将数字拆分为奇数、偶数两组
统计奇数、偶数的个数
3）输出：
第一行：奇数个数 + 空格 + 偶数个数
第二行：所有奇数
第三行：所有偶数

```python
# 传统循环方法
nums = list(map(int, input().split()))

odd_list = []
even_list = []

for num in nums:
    if num % 2 == 1:
        odd_list.append(num)
    else:
        even_list.append(num)

print(len(odd_list), len(even_list))
print(' '.join(map(str, odd_list)))
print(' '.join(map(str, even_list)))
```

【程序填空】根据密码表将密文解密为明文。如，密文“agKxKaKa”可解密为“20151212”。
解密方法：预先约定好一组字母密码存放在元组code密码表中，code[0]~code[9]分别表示数字“0”~“9”对应的密码。输入密文（Ciphertext）并回车（输入字母“q”将退出程序），根据密码表转换成明文（Plaintext，码表中无法转换的码用“？”代替），输出。

```python
code = ('g', 'K', 'a', 'P', 'W', 'x', 'E', 'Q', 'f', 't')
while True:
    d = ''
    s = input("Input the ciphertext('q' for exit):\n")
    if (s == 'q'):
        break
    # (1) 遍历密文的每个字符索引
    for i in range(len(s)):
        # (2) 判断字符是否在密码表中
        if s[i] in code:
            p = code.index(s[i])
            # (3) 将对应数字转为字符串，追加到明文
            d += str(p)
        else:
            d += '?'
    # (4) 输出明文
    print("The Plaintext is :\n" + d)
```

挑战题：词频统计
编写程序，对一段英文文本，统计其中所有不同单词的个数，以及词频最大的前10%的单词；
输入给出一段非空文本，最后以符号#结尾。输入保证存在至少10个不同的单词；
在第一行中输出文本中所有不同单词的个数。注意“单词”不区分英文大小写。

```python
# 1. 读入一整行，直到遇到 #
line = input()
text = line.split('#')[0]

# 2. 按空格分单词，转小写
clean_text = ""
for ch in text:
    if ch.isalnum():          # 字母/数字保留
        clean_text += ch.lower()  # 转小写
    else:
        clean_text += " "     # 其他符号变空格
words = clean_text.split()  # 按空格切出单词列表

# 3. 用字典统计词频
counts = dict()
for w in words:
    counts[w] = counts.get(w, 0) + 1

# 4. 输出：不同单词总数
print(len(counts))

# 5. 排序：先按次数从大到小，再按字母从小到大
# 用列表存 (次数, 单词)
sorted_list = []
for w in counts.keys():
    sorted_list.append((-counts[w], w))  # 负号：排序时降序

sorted_list.sort()

# 6. 取前 10%（至少1个）
n = len(sorted_list)
top_n = max(1, int(n * 0.1))

# 7. 输出：次数:单词
for i in range(top_n):
    neg_cnt, w = sorted_list[i]
    print(f"{-neg_cnt}:{w}")
```

```python
from PIL import Image

# 题目给的字符集（按亮度从高到低排列）
ascii_char = list(r'''$%_&WM#*oahkbdpqwmZO0QLCJUYXzcvunxrjft/\|()1{}[]?-/+@<>i!;:,`'.''')

# 彩色转灰度公式
def get_char(r, g, b, alpha=256):
    if alpha == 0:
        return ' '
    gray = int(r * 0.2126 + g * 0.7152 + b * 0.0722)
    # 把灰度值映射到字符集索引
    idx = int(gray / 256 * len(ascii_char))
    return ascii_char[idx]

# 图片路径
img_path = r"C:\Users\Xu's\OneDrive\python课\05 文件-birdnest.jpg"
# 输出的字符画文本路径
output_path = r"C:\Users\Xu's\OneDrive\python课\birdnest_ascii.txt"

# 调整图片尺寸，避免字符画太宽
width = 120
height = 60

# 打开并处理图片
img = Image.open(img_path).convert("RGB")
img = img.resize((width, height))

# 逐像素生成字符画
text = ""
for y in range(height):
    for x in range(width):
        r, g, b = img.getpixel((x, y))
        text += get_char(r, g, b)
    text += "\n"

# 保存到文件
with open(output_path, "w", encoding="utf-8") as f:
    f.write(text)

print("字符画生成完成，已保存到：", output_path)
```

```python
text = "the clown ran after the car and the car ran into the tent and the tent fell down on the clown and the car"
words = text.split()
dic = {}
for word in words:
    dic[word] = dic.get(word,0) + 1   #词频加1
items = list(dic.items()) #所有数据项组成的列表

items.sort(key = lambda x: x[1], reverse=1) #按词频降序排
for item in items:
    word, count = item
    print('{} = {}'.format(word, count))
```

## 5. 递归函数
规则：函数内部调用自身
必备：递归出口（终止条件）+ 递归公式

【例】计算 Fibonacci 数列第15项的值。

```python
# 用字典pre保存已经算过的结果
pre = {0: 1, 1: 1}

def fib(n):
    if n in pre:
        return pre[n]
    else:
        # 递归计算
        newvalue = fib(n-1) + fib(n-2)
        # 把结果存进字典
        pre[n] = newvalue
        return newvalue

print(fib(15))
```

【例】汉诺塔问题

```python
cnt = 0
def move(src, dst):
    global cnt  # 声明使用全局变量cnt
    cnt += 1    # 每移动一次计数+1
    print('%d: %c -> %c' % (cnt, src, dst))  # 打印移动轨迹

def hanoi(n, src, tmp, dst):
    if n == 1:
        move(src, dst)  # 只剩1个盘，直接从源移到目标
    else:
        hanoi(n-1, src, dst, tmp)  # 步骤1：n-1个从src经dst中转到tmp
        move(src, dst)              # 步骤2：最大盘从src移到dst
        hanoi(n-1, tmp, src, dst)  # 步骤3：n-1个从tmp经src中转到dst

hanoi(3, 'A', 'B', 'C')
print('total = %d' % cnt)
```

例：使用函数求素数和
prime(p)，返回True表示p是素数，返回False表示p不是素数；
PrimeSum(m,n)，函数返回素数和；
其中，两个素数由用户输入

```python
# 定义判断素数的函数
def prime(p):
    # 小于2的数一定不是素数
    if p < 2:
        return False
    if p == 2:
        return True
    # 偶数（除2外）不是素数
    if p % 2 == 0:
        return False
    
    for i in range(3, int(p ** 0.5) + 1, 2):
        if p % i == 0:
            return False
    return True

# 定义求区间内素数和的函数
def PrimeSum(m, n):
    sum_result = 0
    # 遍历 m 到 n 的所有整数
    for num in range(m, n + 1):
        if prime(num):   # 调用素数判断函数
            sum_result += num
    return sum_result

m = int(input("请输入第一个数字："))
n = int(input("请输入第二个数字："))

# 计算并输出结果
total = PrimeSum(m, n)
print(f"区间 [{m}, {n}] 内所有素数的和为：{total}")
```

例：假设你要为学校教务系统开发一个小模块，请创建两个Python文件，完成模块的封装与调用。
创建工具模块hdu_pwd_utils.py，要求：
generate_pwd(length=8)：生成一个指定长度的随机密码（包含大小写字母和数字）；
check_strength(pwd)：检查密码强度，如果长度大于等于8且同时包含字母和数字，返回 True，否则返回 False；
在文件底部，编写测试代码打印一个随机密码。注意：当这个文件被其他脚本import时，测试代码绝对不能自动执行。

```python
import random
import string

# 生成随机密码，默认长度8
def generate_pwd(length=8):
    # 字符池：大小写字母 + 数字
    char_pool = string.ascii_letters + string.digits
    # 随机选取字符并拼接成密码
    pwd = ''.join(random.sample(char_pool, length))
    return pwd

# 检查密码强度
def check_strength(pwd):
    # 判断长度、是否包含字母、数字
    len_flag = len(pwd) >= 8
    letter_flag = any(c.isalpha() for c in pwd)
    digit_flag = any(c.isdigit() for c in pwd)
    return len_flag and letter_flag and digit_flag

# 模块自身测试代码
if __name__ == "__main__":
    test_pwd = generate_pwd()
    print("【模块自测】生成密码：", test_pwd)
```

```python
# 获取用户输入密码长度
try:
    pwd_len = int(input("请输入密码长度："))
except:
    print("输入无效，使用默认长度 8")
    pwd_len = 8

# 调用工具函数生成密码
new_pwd = generate_pwd(pwd_len)
print("生成的密码：", new_pwd)

# 检测密码强度
is_strong = check_strength(new_pwd)
if is_strong:
    print("密码强度：合格")
else:
    print("密码强度：不合格")
```

【例】 继承多个父类。

```python
class Scale:
    def check(self):
        if self.count_person > 500:            
            return "%s是个大公司." %self.name        
        else:            
            return "%s是个小公司." %self.name
class Detail:
    def show(self, scale):
        print("%s 公司有%s名员工."%(scale, self.count_person))
class Company(Scale, Detail):
    def __init__(self, name, count):
        self.name = name
        self.count_person = count
if __name__ == "__main__":
    my_company = Company("ABC", 800)    
    company_scale = my_company.check()
    my_company.show(company_scale)
```

设计学生、本科生、研究生三个类，其中，Student类是父类，包含学生的基本信息，如姓名、平均成绩等；Undergraduate和Postgraduate是其子类：
（1）设计子类的特有属性及输出其所有属性的方法，比如toString() ；
（2）设计一个函数，输入参数是学生列表，输出是每个学生的所有属性。

```python
# 父类：学生
class Student:
    def __init__(self, name, avg_score):
        self.name = name
        self.avg_score = avg_score
    def toString(self):
        return f"姓名：{self.name}，平均分：{self.avg_score}"

# 子类：本科生
class Undergraduate(Student):
    def __init__(self, name, avg_score, major):
        super().__init__(name, avg_score)  # 调用父类构造方法
        self.major = major  # 子类独有属性
    def toString(self):
        return f"本科生-{self.name}，平均分：{self.avg_score}，专业：{self.major}"

# 子类：研究生
class Postgraduate(Student):
    def __init__(self, name, avg_score, tutor):
        super().__init__(name, avg_score)
        self.tutor = tutor  # 子类独有属性
    def toString(self):
        return f"研究生-{self.name}，平均分：{self.avg_score}，导师：{self.tutor}"

# 遍历学生列表，输出所有信息
def show_student_list(student_list):
    for stu in student_list:
        print(stu.toString())

# 测试
if __name__ == "__main__":
    s1 = Undergraduate("张三", 85, "计算机")
    s2 = Postgraduate("李四", 90, "王教授")
    stu_list = [s1, s2]
    show_student_list(stu_list)
```
